## TASK - 1: Create Your Personal Matrix

```python
# TASK - 1
# Student ID: 0432320005101106
# Last two digits = 06
# d1 = last digit, d2 = second last digit

import numpy as np

d1 = 6
d2 = 0

A = np.array([[d1 + 2, d2 + 1],
              [2 * d1, d2 + 2]])

print("Matrix A:")
print(A)
```

## TASK - 2: Analyze Matrix A

```python
# TASK - 2: Analyze Matrix A

print("Shape of A:", A.shape)

det_A = np.linalg.det(A)
print("Determinant of A:", det_A)

rank_A = np.linalg.matrix_rank(A)
print("Rank of A:", rank_A)

eigenvalues_A, eigenvectors_A = np.linalg.eig(A)
print("Eigenvalues of A:", eigenvalues_A)
print("Eigenvectors of A:\n", eigenvectors_A)

if det_A != 0:
    inverse_A = np.linalg.inv(A)
    print("Inverse of A:\n", inverse_A)
else:
    print("Inverse does not exist (determinant is 0)")
```

## TASK - 3: Value Change Experiment

```python
# TASK - 3: Value Change Experiment
# I am changing element A[0,0] by adding +1
# Old value of A[0,0] = 8
# New value of B[0,0] = 9

B = A.copy()
B[0, 0] = B[0, 0] + 1

print("Changed element: A[0,0]")
print("Old value:", A[0, 0])
print("New value:", B[0, 0])
print("\nMatrix B:")
print(B)
```

## TASK - 4: Re-analyze Matrix B

```python
# TASK - 4: Analyze Matrix B

print("Shape of B:", B.shape)

det_B = np.linalg.det(B)
print("Determinant of B:", det_B)

rank_B = np.linalg.matrix_rank(B)
print("Rank of B:", rank_B)

eigenvalues_B, eigenvectors_B = np.linalg.eig(B)
print("Eigenvalues of B:", eigenvalues_B)
print("Eigenvectors of B:\n", eigenvectors_B)

if det_B != 0:
    inverse_B = np.linalg.inv(B)
    print("Inverse of B:\n", inverse_B)
else:
    print("Inverse does not exist (determinant is 0)")
```
