## Problem Statement

**Perform a two-sample t-test on the following data:**

* **Data 1:** \[18.4, 19.1, 17.9, 18.7, 18.3]
* **Data 2:** \[20.2, 20.5, 20.1, 20.3, 20.4]

Use significance level **α = 0.05** to determine whether the population means differ.

### Tasks

1. State $H_0$ and $H_1$. You will assume **equal variances** (pooled).
2. Compute the **sample means**, **sample variances (ddof = 1)**, and the **mean difference**.
3. Calculate the **t-statistic**, **degrees of freedom**, and the **two-tailed p-value**.
4. Construct the **95% confidence interval** for the mean difference.
5. **Decision & Interpretation:** Conclude (reject / fail to reject $H_0$) and explain the result **in one or two plain-English sentences**.



---




## **SOLUTION**


### 1. State $H_0$ and $H_1$. You will assume equal variances (pooled).


*   **Null Hypothesis ($H_0$)**: $\mu_1 = \mu_2$ (The population means of Data 1 and Data 2 are equal.)
*   **Alternative Hypothesis ($H_1$)**: $\mu_1 \neq \mu_2$ (The population means of Data 1 and Data 2 are different.)

### Import Necessary Libraries

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

### Define the two independent datasets

```python
data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])
```

### 2. Compute the sample means, sample variances (ddof = 1), and the mean difference.

```python
# mean difference
mean_diff = np.mean(data1) - np.mean(data2)

# sample sizes
n1, n2 = len(data1), len(data2)

# sample variances
s1_sq, s2_sq = np.var(data1, ddof=1), np.var(data2, ddof=1)
```

### 3. Calculate the t-statistic, degrees of freedom, and the two-tailed p-value.

```python
# pooled variance
sp_sq = ((n1 - 1) * s1_sq + (n2 - 1) * s2_sq) / (n1 + n2 - 2)

# standard error using pooled variance
std_err = np.sqrt(sp_sq * (1/n1 + 1/n2))

# t-statistic
t_statistic = mean_diff / std_err

# degrees of freedom
df = n1 + n2 - 2

# two-tailed p-value
p_value = 2 * stats.t.sf(abs(t_statistic), df)
```

### 4. Construct the 95% confidence interval for the mean difference.

```python
ci = stats.t.interval(0.95, df, loc=mean_diff, scale=std_err)
```

### Display results

```python
print(f"Sample 1 Mean: {np.mean(data1):.2f}")
print(f"Sample 2 Mean: {np.mean(data2):.2f}")
print(f"Sample 1 Variance: {s1_sq:.4f}")
print(f"Sample 2 Variance: {s2_sq:.4f}")
print(f"Mean Difference: {mean_diff:.2f}")
print(f"t-Statistic: {t_statistic:.2f}")
print(f"Degrees of Freedom: {df:.2f}")
print(f"p-Value: {p_value:.4f}")
print(f"Confidence Interval of Mean Difference: [{ci[0]:.2f}, {ci[1]:.2f}]")
```

### 5. Decision & Interpretation


*   **Decision**: Since the p-value (0.0000) is less than the significance level $\alpha$ (0.05), we **reject the null hypothesis**.
*   **Interpretation**: There is statistically significant evidence to conclude that the population means of the two datasets are different. The 95% confidence interval for the mean difference ([-2.31, -1.33]) does not include zero, further supporting this conclusion.
